```python
fruit = ['apple', 'banana', 'cherry', 'durian', 'elderberry']
print(fruit[1])
print(fruit[-2])
print(fruit[:2:1])
print(fruit[3::1])
print(fruit[1:4:1])
fruit.append("Orange") # append adds an element to the end of a list
print(fruit)
fruit.remove("Orange") # it removes the element that is specified 
print(fruit)
fruit.pop()
print(fruit)
```
